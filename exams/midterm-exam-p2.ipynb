## Part 2. (20 points) Gravity Assist Mission to Titan

This problem consists of developing your own standalone python module to simulate the trajectory of a **space probe performing a gravitational slingshot (gravity assist) manoeuvre**. The module should accept initial parameters from the user and deliver customised simulations of two-body systems where the interaction between them is of gravitational nature.

We will analyse how a probe travelling through the Solar System could gain velocity during close flybys of different celestial bodies on its way toward **Saturn's moon Titan**. We will not simulate the full interplanetary trajectory from Earth to Titan. Instead, we will model the **local flyby dynamics** of the probe around a single celestial body using a two-body approximation.

For reference, the scales used in this problem correspond approximately to Voyager-style gravity assist flybys. You can find more information here: https://science.nasa.gov/learn/basics-of-space-flight/primer/

### Physical model

We assume that the most massive object of mass $M$ is a **celestial body used for a gravity assist manoeuvre**, and it is located at the origin of the Cartesian coordinate system $(x,y)$, while the other object is a **space probe** with negligible mass $m\ll M$ moving under its gravitational influence.

In this coordinate system, the position of the probe is $\vec{r} = x\hat{x} + y\hat{y}$, which is the vector pointing from the celestial body to the probe.

The **ODE system** describing the motion of the probe is:

$$\frac{d\vec{r}}{dt}=\vec{v}$$

$$m\frac{d\vec{v}}{dt}=-\frac{G\,m\,M}{r^3}\vec{r}$$

Note that $m$ cancels out in the above equation.

In astrophysics it is common to work in special units when studying gravitational motion:

- distances are typically expressed in astronomical units ($\rm AU$)
- time is expressed in years ($\rm yr$)
- masses are expressed in solar masses ($M_{\odot}$)

### Initial conditions

We initialise the spacecraft far from the celestial body with a fixed incoming velocity.

At $t=0$, we place the probe at:

$$
x_0 = -x_{\rm start}
$$

$$
y_0 = b
$$

with initial velocity

$$
v_{x0} = v_0
$$

$$
v_{y0} = 0
$$

where $b$ is the so-called impact parameter. Thus, the probe approaches the celestial body from the left with a fixed velocity $v_0$.

All simulations should use the **same initial position and velocity**, so that the effect of the celestial body's mass can be compared directly.

Use the following parameters:

- $x_{\rm start} = 0.3\,\rm AU$
- $b = 0.010\,\rm AU$
- $v_0 = 1.7\,\rm AU\,yr^{-1}$

These values correspond to a spacecraft travelling at roughly **8 km/s**, typical of interplanetary missions.
